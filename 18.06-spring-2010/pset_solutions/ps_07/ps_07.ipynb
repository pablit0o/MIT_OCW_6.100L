```python
import numpy as np 
import scipy as sp 
```

## Challenge Problem

```python
def v(n):
    arr = []
    for i in range(n):
        if i == 1 or i == n-1:
            arr.append(1)
            continue
        arr.append(0)
    return arr

for i in range(5,13):
    matrix = sp.linalg.toeplitz(v(i))
    print(f"matrix:\n{matrix}\ndet={sp.linalg.det(matrix)}")
```

```python
def cofactor(m):
    return np.linalg.inv(m).T * np.linalg.det(m)

for i in range(5,13):
    try: # in case of singular matrices
        matrix = sp.linalg.toeplitz(v(i))
        print(f"matrix:\n{matrix}\ndet={sp.linalg.det(matrix)}\ncofactor=\n{cofactor(matrix)}")
    except:
        print(f"matrix:\n{matrix}\ndet={sp.linalg.det(matrix)}\ncofactor=singular")
```
