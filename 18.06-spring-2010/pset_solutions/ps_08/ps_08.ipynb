```python
a = np.array([[.4, .2, .3], [.2, .4, .3], [.4, .4, .4]])
print(sp.linalg.eigvals(a))
```

So about 1, 0.2, and 0

```python
u = [np.array([1, 0, 0]), np.array([100, 0, 0])]

x = 100

for j in range(0,2):
    res = u[j]
    for i in range(x):
        res = np.linalg.matmul(a, res)
    print(f"{u[j]} after {x} matmuls' {res}\n")
```
