# Section 4.4 Problem 35

```python
A = np.eye(4) - np.diag([1, 1, 1], -1)

Q, R = np.linalg.qr(A)

for i in range(100): # surely 100 should be enough
    Qt = Q * math.sqrt(i+1)
    print(f"i={i+1}\nQ={Qt}")
```

from brute force: sqrt(2), sqrt(6), sqrt(3), sqrt(4)=2

```python
Q[:, 0] = Qt[:, 0] * math.sqrt(2)
Q[:, 1] = Qt[:, 1] * math.sqrt(6)
Q[:, 2] = Qt[:, 2] * math.sqrt(3)
Q[:, 3] = Qt[:, 3] * math.sqrt(4)
print(Q)
```

# Section 5.1 Problem 31

```python
for i in range(10):
    H = sp.linalg.hilbert(i+1)
    print(np.linalg.det(H))

H = sp.linalg.hilbert(5)
P, L, U = sp.linalg.lu(H)

print(U.round(3))
```

# Section 5.1 Problem 32

```python
n = [50, 100, 200, 400]
m = 100

# rand
for i in n:
    sum = 0
    for x in range(m): 
        A = np.random.randn(i, i)
        sum += np.linalg.det(A)

    print(f"n={i}; avg det={sum/m}")
```

Note: The results themselves seem to be random, but generally fall under the same scale.
